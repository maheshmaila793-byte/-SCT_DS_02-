```python


# Task 2: Data Cleaning & EDA on Titanic dataset

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Upload dataset
uploaded = files.upload()
df = pd.read_csv("task_2dataset.csv")   # <-- use your uploaded filename

# 1. Basic info
print("Dataset Info:")
print(df.info())
print("\nFirst 5 rows:")
print(df.head())

# 2. Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].median())        # Fill Age with median
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])  # Fill Embarked with mode
if "Cabin" in df.columns:
    df = df.drop(columns=["Cabin"])                     # Drop Cabin

# 3. Encode categorical variables
df["Sex"] = df["Sex"].map({"male":0, "female":1})
df["Embarked"] = df["Embarked"].map({"S":0, "C":1, "Q":2})

# 4. Summary statistics
print("\nSummary Statistics:")
print(df.describe())

# 5. Visualizations
plt.figure(figsize=(8,6))
plt.hist(df["Age"], bins=20, color="skyblue", edgecolor="black")
plt.title("Age Distribution")
plt.show()

plt.figure(figsize=(6,5))
sns.barplot(x="Sex", y="Survived", data=df, hue="Sex", palette="pastel", legend=False)
plt.title("Survival Rate by Gender")
plt.xticks([0,1], ["Male","Female"])
plt.show()

plt.figure(figsize=(6,5))
sns.barplot(x="Pclass", y="Survived", data=df, hue="Pclass", palette="muted", legend=False)
plt.title("Survival Rate by Passenger Class")
plt.show()

plt.figure(figsize=(6,5))
sns.barplot(x="Embarked", y="Survived", data=df, hue="Embarked", palette="Set2", legend=False)
plt.title("Survival Rate by Embarked Port")
plt.xticks([0,1,2], ["S","C","Q"])
plt.show()

# 6. Correlation heatmap (numeric only)
numeric_df = df.select_dtypes(include=["number"])
plt.figure(figsize=(10,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
